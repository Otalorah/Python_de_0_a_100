# Sintaxis

La sintaxis de Python como lenguaje es bastante sencilla en general, dejare algunas funciones basicas para que tomen la idea general de como se escribe codigo dentro de Python.
A diferencia de otros lenguajes, **Python prioriza la legibilidad**. 
En lugar de utilizar llaves {} para definir los bloques de código o exigir un punto y coma ; al final de cada instrucción, Python utiliza la indentación (espacios en blanco al inicio de la línea) para estructurar la lógica. Esto hace que el código sea muy limpio y fácil de leer.

```python
nombre = "Ana" 
edad = 25
altura = 1.68
es_estudiante = True

print("Hola, mi nombre es", nombre)

print("Contando hasta 3:")
for numero in range(3):
    print(numero)

def saludar_usuario(usuario):
    mensaje = "¡Bienvenido al sistema, " + usuario + "!"
    return mensaje

saludo_final = saludar_usuario(nombre)
print(saludo_final)
```

### PEP 8 (Estándar)

Las siglas significan Python Enhancement Proposal. Son puntos clave que hacen que el código dentro de Python se vea mucho más organizado. 
La razón principal es que Python, al ser tan sencillo de escribir, puede generar ciertos problemas de lectura para personas externas que no logran comprender la lógica detrás de algún programa si no hay un orden común.

- **Indentación:** Tener en cuenta que esto es sumamente **importante**, debido a que es la base de la "organización" y estructura del código dentro de Python. Al no existir llaves `{}` para abrir o cerrar bloques, Python usa los espacios. La convención estricta es usar **4 espacios** por cada nivel.
- **Longitud máxima de línea:** Se recomienda no pasar de los **79 caracteres** por línea. Esto permite leer el código fácilmente en pantallas más pequeñas o al poner dos archivos uno al lado del otro, sin necesidad de hacer *scroll* horizontal.
- **Nombres de variables y funciones (`snake_case`):** Deben escribirse completamente en minúsculas. Si el nombre está compuesto por varias palabras, se separan con un guion bajo.
  - *Correcto:* `nota_final = 10`
  - *Incorrecto:* `notafinal = 10`
- **Constantes:** Las variables que representan valores fijos (que no deben cambiar a lo largo del programa) se escriben totalmente en mayúsculas, separando las palabras con guiones bajos.
  - *Ejemplo:* `LIMITE_ASISTENCIAS = 3`
- **Espacios en blanco con operadores:** Siempre se debe dejar un espacio antes y después de signos como `=` (asignación) y operadores matemáticos (`+`, `-`, `*`, `/`). Esto evita que el código se vea amontonado.
  - *Correcto:* `x = y + 5`
  - *Incorrecto:* `x=y+5`
- **Líneas en blanco para dar "respiro":** Se recomienda dejar dos líneas en blanco antes de definir una función nueva. Dentro de la misma función, se pueden usar líneas en blanco simples para separar bloques lógicos de código, como si fueran los párrafos de un libro.
- **Comentarios precisos:** Se escriben utilizando el símbolo `#`. Deben utilizarse para explicar *por qué* se está haciendo algo, no *qué* se está haciendo. Si las variables están bien nombradas, el "qué" debe ser evidente.

```python

# 1. Constantes: Totalmente en mayúsculas
IMPUESTO_NACIONAL = 0.19
LIMITE_DESCUENTO = 100000

# 2. Variables: snake_case (minúsculas con guiones bajos)
nombre_cliente = "Ana Gómez"
total_carrito = 150000


# 3. Líneas en blanco: Dos líneas vacías antes de definir una función
def calcular_total_factura(subtotal):
    # 4. Indentación: Exactamente 4 espacios para el bloque interno
    
    # 5. Espacios con operadores: Un espacio antes y después de =, *, y +
    valor_impuesto = subtotal * IMPUESTO_NACIONAL
    total_parcial = subtotal + valor_impuesto
    
    # 6. Comentarios de valor: Explican el "por qué", no el "qué"
    # Se aplica el descuento solo si supera el límite establecido por gerencia
    if total_parcial > LIMITE_DESCUENTO:
        total_parcial = total_parcial - 5000
        
    return total_parcial


# Llamado a la función y muestra en consola
precio_final = calcular_total_factura(total_carrito)
print("El total a pagar por", nombre_cliente, "es de:", precio_final)
```

## Tipos de Datos Primitivos, Operadores Aritmeticos y de Asignacion

### Operadores de Asignacion y Aritmeticos
Estos operadores funcionan principalmente para dar valores a "algo", entonces nos ayudan a poner informacion dentro de las variables

- =     | Asigna valores a variables    
- +=    | Es la formula equivalente de " algo mas algo"
- -=    | Es la formula equivalente de " algo menos algo"
- *=    | Es la formula equivalente de " algo multiplicado por algo"
- /=    | Es la formula equivalente de " algo dividido por algo"
- //=   | Es la formula equivalente de " algo dividido enteramente por algo" (descarta la parte decimal de cualquier valor; **ejemplo**: 5//2 = 2. 5 divido 2 = 2.5, pero descartamos parte decimal, entonces el resultado es 2)
- %=    | Es la formula equivalente de " algo modulo algo"
- **=   | Es la formula equivalente de " algo potencia de algo"

```python
x = 5
y = 6

# Vamos a sumar 4 al 5 | x | de manera iterativa
for _ in range(2):
    x += 2

print("Valor de x tras la suma iterativa:", x)

# Vamos a potenciar 3 veces de manera iterativa al 6 | y |
for _ in range(2):
    y **= 3

print("Valor de y tras la potenciación iterativa:", y)

# Vamos a restar 2 al | y | de manera iterativa
for _ in range(2):
    y -= 1
    
print("Valor de y tras la resta iterativa:", y)

# Multiplicamos la | x | por 20
x *= 20

print("Valor de x tras multiplicar por 20:", x)

# Sacamos la parte entera de y en una division entre 5.6
y //= 5.6

print("Valor de y tras la división entera:", y)

# Sacamos modulo 1231 de x
x %= 1231

print("Valor final de x tras el módulo:", x)
```

## int, float, complex
Representan directamente cualquier tipo de numero que tenemos dentro de nuestro sistema
1. int: enteros, sea int un valor numerico, entonces int pertenece a los Naturales
2. float: Reales, sea float un valor numerico, entonces float pertenece a los Reales
3. complex: complejos, sea complex un valor numerico, entonces complex pertenece a los complejos

```python
# 1. Tipos Principales 
entero = 42
flotante = 3.14
complejo = 3 + 5j

# ========================================== COMPLEX ========================================================== #
# Extracción de partes en complejos
# Permite acceder por separado a la parte real e imaginaria de un número complejo.
parte_real = complejo.real
parte_imag = complejo.imag


# ========================================== Constructores y especiales ========================================================== #

# inf, -inf y nan, representan valores extremadamente grandes, por su nombre, infinito (-,+) o valores NO numericos nan
c_str_float = float("inf")

# Cuando pasamos de un valor numerico a otro, generalmente le quitamos o agregamos a la variable original.
# "truncar": un valor el cual hace referencia a cortar, en este caso, cortamos el 9 y dejamos el 3
entero_desde_float = int(3.9)
print(entero_desde_float)

# Para un paso de int a float, siempre se le agrega un x.0
entero_a_float = float(4)
print(entero_a_float)

# ========================================== Aritmetica ========================================================== #

# Si combinas int y float, el int se convierte a float.
# Si combinas con complex, el otro se convierte a float o complex.
# Si combinas int con complex, este pasa a complex

resultado_mixto = 2 + 4.5
print(resultado_mixto) 
resultado_mixto = 2.5 + complejo
print(resultado_mixto) 
resultado_mixto = 2 + complejo
print(resultado_mixto) 


# ========================================== Ejemplos Aritmeticos ========================================================== #

x, y = 10, 3

suma = x + y                # 13
resta = x - y               # 7
multiplicacion = x * y      # 30
division = x / y            # 3.3333333333333335
division_entera = x // y    # 3 (redondea hacia menos infinito)
modulo = x % y              # 1 (residuo)
potencia = x ** y           # 1000 (o pow(x, y))

negativo = -x               # -10
positivo = +x               # 10
absoluto = abs(-5)          # 5
par_divmod = divmod(x, y)   # (3, 1) -> Equivale a (x // y, x % y)
conjugado = (3 + 5j).conjugate() # 3 - 5j (solo para complejos)
```
