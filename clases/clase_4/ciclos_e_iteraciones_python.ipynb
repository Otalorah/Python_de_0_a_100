## 1. El Bucle `while`

### ¿Qué es?
El bucle `while` es una estructura de control que repite un bloque de código **mientras una condición lógica se evalúe como verdadera (`True`)**.

Antes de cada repetición, Python evalúa la condición:
- Si es `True`, ejecuta las instrucciones dentro del bloque indentado.
- Si es `False`, detiene inmediatamente el ciclo y transfiere el flujo a la siguiente línea del programa.

#### Componentes indispensables de un `while`:
1. **Estado inicial:** Definición de una o más variables antes de ingresar al bucle.
2. **Condición de frontera:** Expresión de comparación que determina si el ciclo continúa o finaliza.
3. **Paso de actualización:** Modificación dentro del cuerpo del bucle que altera las variables de control para asegurar que eventualmente la condición sea `False` y evitar un **bucle infinito**.

```python
# Ejemplo 1.1: Contador y acumulador básico
contador = 1
limite = 5
suma_acumulada = 0

while contador <= limite:
    suma_acumulada += contador
    print(f"Iteracion: {contador} | Suma acumulada: {suma_acumulada}")
    contador += 1  # Actualizacion del estado

print(f"-> Fin del ciclo. Suma total de 1 a {limite}: {suma_acumulada}")
```

```python
# Ejemplo 1.2: Reduccion matematica mediante division entera sucesiva
# Descomposicion de un numero en sus digitos individuales sin convertir a texto
numero = 4529
temporal = numero
suma_digitos = 0

print(f"Descomponiendo {numero} digito a digito con divisiones sucesivas:")
while temporal > 0:
    ultimo_digito = temporal % 10
    suma_digitos += ultimo_digito
    print(f"Digito extraido: {ultimo_digito} | Valor restante: {temporal // 10}")
    temporal = temporal // 10  # Reduce el numero un orden de magnitud

print(f"-> Suma de todos los digitos de {numero}: {suma_digitos}")
```

## 2. El Bucle `for` y la Función `range`

### ¿Qué es?
En Python, el bucle `for` está diseñado para **recorrer secuencialmente colecciones de elementos u objetos iterables** (como cadenas de texto, rangos numéricos, listas o tuplas).

En lugar de manipular manualmente un índice numérico o una variable de incremento, el bucle `for` toma un elemento del iterable en cada paso y lo asigna a una variable de iteración hasta agotar la secuencia.

### La función nativa `range()`
`range()` genera una secuencia aritmética inmutable de enteros bajo demanda, manteniendo un uso de memoria constante y mínimo:

- `range(fin)`: Genera enteros desde `0` hasta `fin - 1` con paso de `1`.
- `range(inicio, fin)`: Genera enteros desde `inicio` hasta `fin - 1` con paso de `1`.
- `range(inicio, fin, paso)`: Genera enteros desde `inicio` hasta `fin - 1` avanzando según el `paso` indicado (admite valores negativos para recorridos decrecientes).

```python
# Ejemplo 2.1: Variantes de la funcion range()
print("1. range(stop):")
for i in range(4):
    print(f"i = {i}", end=" | ")
print()

print("\n2. range(start, stop):")
for i in range(3, 8):
    print(f"i = {i}", end=" | ")
print()

print("\n3. range(start, stop, step) regresivo:")
for i in range(10, 0, -2):
    print(f"i = {i}", end=" | ")
print()
```

```python
# Ejemplo 2.2: Iteracion directa sobre cadenas de texto (secuencias nativas)
frase = "Python 0 a 100"
conteo_vocales = 0
conteo_consonantes = 0

for caracter in frase:
    caracter_minuscula = caracter.lower()
    if caracter_minuscula in "aeiou":
        conteo_vocales += 1
    elif caracter_minuscula.isalpha():
        consonantes_temporales = conteo_consonantes + 1
        conteo_consonantes = consonantes_temporales

print(f"Texto evaluado: '{frase}'")
print(f"Total vocales encontradas: {conteo_vocales}")
print(f"Total consonantes encontradas: {conteo_consonantes}")
```

## 3. Sentencias de Alteración de Flujo: `break`, `continue` y `pass`

### ¿Qué es?
Son instrucciones nativas que permiten modificar el comportamiento habitual de una iteración en tiempo de ejecución:

- **`break`**: Finaliza inmediatamente el bucle contenedor más próximo. El control del programa salta a la primera línea fuera de dicho ciclo.
- **`continue`**: Interrumpe la iteración actual y salta directamente al encabezado del ciclo para evaluar la siguiente condición o tomar el próximo elemento.
- **`pass`**: Representa una operación nula (*no-op*). Sirve como relleno sintáctico donde se exige una instrucción pero no se desea ejecutar ninguna acción.

```python
# Ejemplo 3.1: Uso de 'continue' para filtrar elementos
# Imprimir impares del 1 al 15 saltando los que sean multiplos de 3
print("Numeros impares no divisibles entre 3:")
for numero in range(1, 16):
    if numero % 2 == 0:
        continue  # Omite los numeros pares
    if numero % 3 == 0:
        continue  # Omite los multiplos de 3
    print(f"Numero admitido: {numero}")
```

```python
# Ejemplo 3.2: Uso de 'break' para busqueda con detencion temprana
limite = 40
objetivo = 28  # Primer numero divisible entre 4 y 7 simultaneamente

print(f"Buscando el primer multiplo comun de 4 y 7 en el rango 1 a {limite}...")
for num in range(1, limite + 1):
    if num % 4 == 0 and num % 7 == 0:
        print(f"-> Objetivo encontrado: {num}. Abortando el ciclo con 'break'.")
        break  
    print(f"Revisando: {num}")

print("Continuando con el flujo principal del programa.")
```

## 4. La Cláusula `else` en Bucles (`for-else` y `while-else`)

### ¿Qué es?
En Python, los bucles `for` y `while` pueden tener un bloque `else` asociado al final.

**Regla fundamental:**
- El bloque `else` **SÍ se ejecuta** cuando el ciclo finaliza de forma natural (es decir, cuando la condición del `while` se convierte en `False`, o cuando el `for` agota todos los elementos del iterable).
- El bloque `else` **NO se ejecuta** si el ciclo fue interrumpido abruptamente mediante una sentencia `break`.

Esta característica resulta útil en algoritmos de búsqueda y verificación, pues evita declarar variables bandera adicionales (`encontrado = False`).

```python
# Ejemplo 4.1: Determinacion de primalidad con for-else
candidato = 29

print(f"Analizando primalidad de: {candidato}")
if candidato <= 1:
    print(f"-> {candidato} no es primo por definicion.")
else:
    for divisor in range(2, candidato):
        if candidato % divisor == 0:
            print(f"-> {candidato} no es primo, es divisible por {divisor}.")
            break 
    else:
        # Se ejecuta solo si el bucle recorrio todos los divisores sin toparse con un break
        print(f"-> {candidato} ES UN NUMERO PRIMO.")
```

## 5. Bucles Anidados

### ¿Qué es?
Un bucle anidado es un ciclo ubicado dentro del bloque de instrucciones de otro ciclo. Por cada iteración individual del bucle exterior, el bucle interior se ejecuta por completo.

> **Nota sobre el alcance:** Una instrucción `break` o `continue` colocada dentro del bucle anidado solo altera el ciclo interno más cercano; no interrumpe el ciclo exterior.

```python
# Ejemplo 5.1: Matriz de multiplicacion formateada
print("Tabla de multiplicar (filas x columnas) del 1 al 5:")
for fila in range(1, 6):
    # 1
    for columna in range(1, 6):
        # 1 * 1, 1 * 2, 1 * 3, 1 * 4, 1 * 5
        # 2 * 1, 2 * 2, 2 * 3, 2 * 4, 2 * 5
        # .
        # .
        # .
        producto = fila * columna
        print(f"{producto:3d}", end=" ")
    print()  # Salto de linea al terminar cada fila
```

## Resumen

| Estructura / Sentencia | ¿Qué hace? | Cuándo utilizarla |
| :--- | :--- | :--- |
| `while` | Repetición condicional indeterminada | Cuando no se conoce previamente el número exacto de iteraciones |
| `for` | Recorrido determinado de secuencias | Cuando se itera sobre colecciones, textos o rangos definidos |
| `range(start, stop, step)` | Generador inmutable de sucesiones de números | Para conteos numéricos directos o definir límites de repetición |
| `break` | Interrupción inmediata del ciclo | Al encontrar una solución anticipada o por motivos de seguridad |
| `continue` | Salto de la iteración actual | Para omitir casos específicos y pasar directo a la siguiente vuelta |
| `else` en bucles | Ejecución si el ciclo finaliza sin interrupción | En algoritmos de búsqueda para verificar que se revisó todo sin `break` |
