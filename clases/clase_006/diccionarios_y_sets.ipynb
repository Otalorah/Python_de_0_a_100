## 1. 📖 Diccionarios (`dict`)

### 1.1 ¿Qué es un Diccionario?
Un **diccionario** en Python es una estructura de datos mutable que almacena información en pares **clave-valor** (`key: value`). 

A diferencia de las listas y tuplas, donde accedemos a los elementos mediante un índice numérico secuencial (`0, 1, 2, ...`), en un diccionario accedemos a los valores a través de su **clave**.

Internamente, Python implementa los diccionarios mediante una **Tabla Hash** (*Hash Map*). Esto significa que el tiempo para buscar, insertar o eliminar un elemento es prácticamente **instantáneo** ($O(1)$ en promedio), sin importar si el diccionario tiene 10 o 1,000,000 de registros.

#### Reglas fundamentales de las claves:
1. **Deben ser únicas:** No pueden existir dos claves idénticas dentro del mismo diccionario. Si se repite una clave al asignar, se sobreescribe el valor anterior.
2. **Deben ser inmutables (*hashables*):** Solo tipos de datos inmutables pueden ser claves (`str`, `int`, `float`, `tuple`, `bool`). Las listas y diccionarios **NO** pueden ser claves.
3. **Los valores pueden ser de cualquier tipo:** Un valor puede ser un entero, un texto, una lista, otro diccionario o cualquier objeto.


````python
# Creación de diccionarios

# 1. Usando llaves {}
perfil_estudiante = {
    "estudiantes" : [{
        "nombre" : "alberto1",
        "carrera" : "Industrial",
        "codigo" : 20212020763
    },
                    {
        "nombre" : "alberto2",
        "carrera" : "Industrial",
        "codigo" : 20212020763
    },
                    {
        "nombre" : "alberto3",
        "carrera" : "Industrial",
        "codigo" : 20212020763
    },
                    {
        "nombre" : "alberto4",
        "carrera" : "Industrial",
        "codigo" : 20212020763
    }]
}

# 2. Usando la función constructora dict()
# configuracion = dict(servidor="localhost", puerto=8080, debug=True)

# configuracion = {
#     "servidor" : "localhost",    
# }
# # 3. Diccionario vacío
# vacio_1 = {}
# vacio_2 = dict()

# print("Perfil estudiante:", perfil_estudiante)
# print("Tipo:", type(perfil_estudiante))
# print("Configuración:", configuracion)
# print("¿vacio_1 es de tipo dict?:", isinstance(vacio_1, dict))

print(perfil_estudiante.get("estudiantes")[2])
````

### 1.2 Acceso, Modificación y Acceso Seguro (`.get()`)

Existen dos formas principales de consultar el valor de una clave:
* **Indexación directa con corchetes `dict[clave]`:** Si la clave existe, retorna el valor. Si la clave **no existe**, Python detiene la ejecución arrojando un error de tipo `KeyError`.
* **Método seguro `.get(clave, valor_por_defecto)`:** Si la clave existe, retorna su valor. Si no existe, retorna `None` (o el valor alternativo que definas) **sin romper el programa**.


````python
# Consulta mediante corchetes vs .get()
estudiante = {"nombre": "Camilo", "semestre": 6}

# Acceso directo
print("Nombre:", estudiante["nombre"])

# ¿Qué pasa si la clave no existe con corchetes?
# print(estudiante["promedio"])  # ❌ Descomenta esta línea para ver el KeyError

# Acceso seguro con .get()
promedio = estudiante.get("promedio")
print("Promedio consultado con get():", promedio)  # Imprime None

# .get() con valor de respaldo por defecto
promedio_seguro = estudiante.get("promedio", 0.0)
print("Promedio con valor de respaldo:", promedio_seguro)

# Asignar o modificar valores
estudiante["semestre"] = 7         # Modificación de una clave existente
estudiante["promedio"] = 4.3       # Inserción de una nueva clave
estudiante["materias"] = ["Redes", "Bases de Datos"]  # El valor puede ser una lista

print("\nEstudiante actualizado:", estudiante)
````

### 1.3 Eliminación de Elementos

Para remover pares clave-valor contamos con:
1. `del dict[clave]`: Elimina la clave directamente. Lanza `KeyError` si no existe.
2. `dict.pop(clave, valor_defecto)`: Elimina la clave y **retorna su valor asociado**. Si no existe y se pasa un segundo argumento, lo devuelve en lugar de arrojar error.
3. `dict.popitem()`: Elimina y retorna el último par clave-valor insertado (orden LIFO).
4. `dict.clear()`: Vacía completamente el diccionario.


````python
inventario = {
    "manzanas": 50,
    "peras": 30,
    "naranjas": 20,
    "uvas": 15
}

# 1. del
del inventario["manzanas"]
print("Tras eliminar manzanas:", inventario)

# 2. pop(): extrae y devuelve el valor
cantidad_peras = inventario.pop("peras")
print(f"Se extrajeron {cantidad_peras} peras. Inventario restante: {inventario}")

# pop() con valor por defecto si la clave no existe
resultado = inventario.pop("mangos", "Producto no disponible")
print("Resultado pop de 'mangos':", resultado)

# 3. popitem()
ultimo = inventario.popitem()
print("Último elemento extraído:", ultimo)
print("Inventario actual:", inventario)
````

### 1.4 Iteración y Métodos Nativos Fundamentales

Los diccionarios proporcionan tres métodos clave para inspeccionar y recorrer sus datos:
- `.keys()`: Retorna una vista de todas las claves.
- `.values()`: Retorna una vista de todos los valores.
- `.items()`: Retorna una vista de tuplas con los pares `(clave, valor)`. Esta es la forma más idiomática y limpia de iterar en Python.


````python
calificaciones = {
    "Carlos": 4.5,
    "Daniela": 4.8,
    "Andrés": 3.9,
    "Karina": 5.0
}

print("--- 1. Iterando sobre las claves (.keys()) ---")
for nombre in calificaciones.keys():
    print(f"Estudiante: {nombre}")

print("\n--- 2. Iterando sobre los valores (.values()) ---")
for nota in calificaciones.values():
    print(f"Nota registrada: {nota}")

print("\n--- 3. Iterando sobre clave y valor simultáneamente (.items()) ---")
for nombre, nota in calificaciones.items():
    estado = "Aprobado" if nota >= 3.0 else "Reprobado"
    print(f"{nombre:<10} -> Nota: {nota:.1f} ({estado})")

# Métodos adicionales útiles:
# Comprobación de existencia rápida (O(1))
print("\n¿Existe Daniela en el registro?:", "Daniela" in calificaciones)
print("¿Existe Pedro en el registro?:", "Pedro" in calificaciones)

# .update(): fusión de diccionarios
nuevos = {"Pedro": 4.2, "Carlos": 4.7}  # Carlos se actualizará
calificaciones.update(nuevos)
print("\nCalificaciones tras .update():", calificaciones)
````

### 💡 Mini-Reto Guiado 1: Conteo de Frecuencias de Palabras
Un patrón sumamente común en programación es contar la frecuencia de aparición de elementos usando un diccionario como acumulador.


````python
texto = "python es genial y aprender python con acm es divertido y util"
palabras = texto.split()

# Conteo manual usando lógica condicional
frecuencias = {}
for p in palabras:
    if p in frecuencias:
        frecuencias[p] += 1
    else:
        frecuencias[p] = 1

print("Frecuencia de palabras (con if/else):")
print(frecuencias)

# Alternativa más concisa y pythónica usando .get()
frecuencias_get = {}
for p in palabras:
    frecuencias_get[p] = frecuencias_get.get(p, 0) + 1

print("\nFrecuencia usando .get():")
print(frecuencias_get)
````

## 2. ⭕ Conjuntos (`set`)

### 2.1 ¿Qué es un Set?
Un **conjunto (`set`)** en Python es una colección **mutable**, **no ordenada** y compuesta exclusivamente por **elementos únicos** (sin duplicados).

Al igual que los diccionarios, los `set` se implementan mediante **tablas hash**. Sin embargo, a diferencia de los diccionarios que guardan pares `clave: valor`, el `set` almacena **solo las claves**.

#### Características clave:
1. **Unicidad:** Si intentas insertar un elemento duplicado, el `set` lo ignora automáticamente.
2. **Sin orden ni índices:** Los elementos no tienen una posición fija. No puedes hacer `mi_set[0]` (arroja `TypeError`).
3. **Elementos inmutables (*hashables*):** Cada elemento del conjunto debe ser inmutable (`int`, `float`, `str`, `tuple`). No se pueden meter listas ni diccionarios dentro de un set.
4. **Cuidado con la sintaxis:** `{}` crea un **diccionario vacío**, no un set. Para crear un set vacío debes usar obligatoriamente `set()`.


````python
# Creación de conjuntos

# 1. Con llaves y elementos
frutas = {"manzana", "banana", "cereza", "manzana"}  # Duplicado intencional
print("Conjunto de frutas (duplicados eliminados automáticamente):", frutas)

# 2. Desde un iterable (útil para desduplicar listas al instante)
numeros_con_duplicados = [1, 2, 2, 3, 4, 4, 4, 5, 1, 3]
numeros_unicos = set(numeros_con_duplicados)
print("Lista original con duplicados:", numeros_con_duplicados)
print("Convertida a set (únicos):", numeros_unicos)

# 3. Conjunto vacío: ¡MUY IMPORTANTE!
no_es_un_set = {}
si_es_un_set = set()

print("Tipo de {}:", type(no_es_un_set))      # <class 'dict'>
print("Tipo de set():", type(si_es_un_set))   # <class 'set'>
````

### 2.2 Métodos Nativos de Manipulación de Sets

- `.add(elem)`: Agrega un elemento al conjunto. Si ya existe, no hace nada.
- `.update(iterable)`: Agrega múltiples elementos desde una lista, tupla u otro set.
- `.remove(elem)`: Elimina un elemento. **Lanza un `KeyError` si el elemento no existe.**
- `.discard(elem)`: Elimina un elemento de forma segura. **Si no existe, no genera error.**
- `.pop()`: Elimina y retorna un elemento arbitrario del conjunto.
- `.clear()`: Elimina todos los elementos del conjunto.


````python
lenguajes = {"Python", "C++", "Java"}

# Agregar elementos
lenguajes.add("Rust")
lenguajes.add("Python")  # Ya existe, no produce cambios ni errores
print("Tras agregar Rust y Python:", lenguajes)

# Actualizar con múltiples elementos
lenguajes.update(["JavaScript", "Go", "C++"])
print("Tras update:", lenguajes)

# Eliminar con discard (seguro)
lenguajes.discard("Java")
lenguajes.discard("Kotlin")
print("Tras discard de Java y Kotlin:", lenguajes)

# Eliminar con remove (precaución)
lenguajes.remove("C++")
print("Tras remove de C++:", lenguajes)

# Pertenencia ultra rápida O(1)
print("¿Python está en el conjunto?:", "Python" in lenguajes)
````

### 2.3 Operaciones de Teoría de Conjuntos (Álgebra de Conjuntos)

Una de las mayores virtudes de los `set` en Python es que implementan de forma nativa los operadores y métodos de la **teoría matemática de conjuntos**:

| Operación Matemática | Notación | Operador en Python | Método en Python | Descripción |
| :--- | :---: | :---: | :---: | :--- |
| **Unión** | $A \cup B$ | `A \| B` | `A.union(B)` | Todos los elementos que están en $A$ o en $B$. |
| **Intersección** | $A \cap B$ | `A & B` | `A.intersection(B)` | Elementos que están simultáneamente en $A$ y en $B$. |
| **Diferencia** | $A \setminus B$ | `A - B` | `A.difference(B)` | Elementos que están en $A$ pero **no** en $B$. |
| **Diferencia Simétrica** | $A \Delta B$ | `A ^ B` | `A.symmetric_difference(B)` | Elementos que están en $A$ o en $B$, pero **no en ambos**. |


````python
# Conjuntos de ejemplo: Habilidades de dos grupos
grupo_backend = {"Python", "SQL", "Docker", "Linux", "Git"}
grupo_frontend = {"JavaScript", "HTML", "CSS", "Git", "Docker"}

# 1. UNIÓN (|): Todas las habilidades disponibles en el equipo
todas_habilidades = grupo_backend | grupo_frontend
print("1. Unión (backend | frontend):")
print("  ", todas_habilidades)

# 2. INTERSECCIÓN (&): Habilidades compartidas por ambos grupos
habilidades_comunes = grupo_backend & grupo_frontend
print("\n2. Intersección (backend & frontend):")
print("  ", habilidades_comunes)

# 3. DIFERENCIA (-): Habilidades exclusivas de backend
solo_backend = grupo_backend - grupo_frontend
print("\n3. Diferencia (backend - frontend):")
print("  ", solo_backend)

# Habilidades exclusivas de frontend
solo_frontend = grupo_frontend - grupo_backend
print("   Diferencia (frontend - backend):")
print("  ", solo_frontend)

# 4. DIFERENCIA SIMÉTRICA (^): Habilidades que solo domina uno de los dos grupos
no_compartidas = grupo_backend ^ grupo_frontend
print("\n4. Diferencia Simétrica (backend ^ frontend):")
print("  ", no_compartidas)
````

### 2.4 Relaciones de Inclusión y Subconjuntos

También podemos comparar si un conjunto está contenido dentro de otro:
- `A.issubset(B)` o `A <= B`: ¿Es $A$ un subconjunto de $B$? (¿Todos los elementos de $A$ están en $B$?)
- `A.issuperset(B)` o `A >= B`: ¿Es $A$ un superconjunto de $B$?
- `A.isdisjoint(B)`: ¿Son $A$ y $B$ disjuntos? (¿No tienen **ningún** elemento en común?)


````python
requisitos_junior = {"Git", "Python"}
perfil_candidato_1 = {"Python", "Git", "SQL", "Docker"}
perfil_candidato_2 = {"JavaScript", "HTML", "CSS"}

# Verificación de subconjuntos
print("¿Candidato 1 cumple todos los requisitos mínimos?:", requisitos_junior.issubset(perfil_candidato_1))
print("¿Candidato 2 cumple todos los requisitos mínimos?:", requisitos_junior.issubset(perfil_candidato_2))

# Verificación de superconjunto
print("¿El perfil 1 contiene todos los requisitos junior?:", perfil_candidato_1.issuperset(requisitos_junior))

# Verificación de disjuntos
print("¿Requisitos y Candidato 2 son totalmente independientes (disjuntos)?:", requisitos_junior.isdisjoint(perfil_candidato_2))
````

## 3. ⚖️ Criterios de Selección en la Estructura de Datos

A menudo, la diferencia entre un programa lento y uno de alto rendimiento radica en la **elección de la estructura de datos adecuada**.

### 3.1 Cuadro Comparativo Integral de Colecciones Nativas

| Estructura | Sintaxis | Ordenada | Indexable | Mutable | ¿Permite Duplicados? | Búsqueda (`x in col`) |
| :--- | :---: | :---: | :---: | :---: | :---: | :---: |
| **Lista (`list`)** | `[a, b, c]` | ✅ Sí | ✅ Sí (`col[i]`) | ✅ Sí | ✅ Sí | $O(n)$ Lineal |
| **Tupla (`tuple`)** | `(a, b, c)` | ✅ Sí | ✅ Sí (`col[i]`) | ❌ No (Inmutable) | ✅ Sí | $O(n)$ Lineal |
| **Conjunto (`set`)** | `{a, b, c}` | ❌ No | ❌ No | ✅ Sí | ❌ **No (Únicos)** | **$O(1)$ Constante** |
| **Diccionario (`dict`)** | `{k: v}` | ✅ Sí (desde Py 3.7) | ❌ Por índice / ✅ Por clave | ✅ Sí | Claves ❌ / Valores ✅ | **$O(1)$ por clave** |

---

### 3.2 ¿Por qué $O(1)$ vs $O(n)$ es tan importante?
- En una **lista**, para saber si un elemento está presente (`x in lista`), Python debe recorrer los elementos uno por uno desde el principio hasta el final en el peor de los casos ($O(n)$). Si la lista tiene 1,000,000 de elementos, puede requerir hasta 1,000,000 de comparaciones.
- En un **set** o en las claves de un **dict**, Python aplica una función matemática (*hash*) directamente al valor buscado para calcular su dirección de memoria y revisarla de forma inmediata ($O(1)$). ¡Tarda prácticamente lo mismo con 10 que con 1,000,000 de elementos!


````python
import time

# Demostración práctica de rendimiento: Búsqueda en Lista vs Set
TAMANO = 200_000
datos_lista = list(range(TAMANO))
datos_set = set(range(TAMANO))
elemento_a_buscar = -999  # Peor caso: el elemento no existe

# 1. Búsqueda en Lista (O(n))
t_inicio = time.perf_counter()
existe_en_lista = elemento_a_buscar in datos_lista
t_fin = time.perf_counter()
tiempo_lista = (t_fin - t_inicio) * 1000  # milisegundos

# 2. Búsqueda en Set (O(1))
t_inicio = time.perf_counter()
existe_en_set = elemento_a_buscar in datos_set
t_fin = time.perf_counter()
tiempo_set = (t_fin - t_inicio) * 1000  # milisegundos

print(f"Búsqueda en Lista ({TAMANO:,} elementos): {tiempo_lista:.4f} ms")
print(f"Búsqueda en Set   ({TAMANO:,} elementos): {tiempo_set:.4f} ms")
if tiempo_set > 0:
    print(f"-> ¡El Set fue aproximadamente {tiempo_lista / max(tiempo_set, 1e-6):.1f} veces más rápido!")
````

### 3.3 Guía de Decisión Rápida: "¿Qué estructura debo usar?"

```
                                  ¿Tus datos vienen en pares Atributo / Valor?
                                                 /            \
                                              (Sí)            (No)
                                              /                  \
                                       [ DICCIONARIO ]      ¿Necesitas que cada elemento sea único
                                                            o realizar operaciones de conjuntos?
                                                                   /             \
                                                                (Sí)             (No)
                                                                /                   \
                                                             [ SET ]          ¿La colección debe ser inmutable
                                                                              (constante, protegida o clave)?
                                                                                     /             \
                                                                                  (Sí)             (No)
                                                                                  /                   \
                                                                              [ TUPLA ]             [ LISTA ]
```

#### Resumen de Casos de Uso:
1. **Usa `list` cuando:**
   - El orden de inserción es relevante y necesitas acceder por posición (`elementos[0]`, `elementos[-1]`).
   - Los elementos se pueden repetir (ej. un registro de temperaturas medidas cada segundo).
   - Necesitas una cola o pila secuencial (`append`, `pop`).

2. **Usa `tuple` cuando:**
   - Representas un registro fijo o coordenadas geométricas inmutables (ej. `(latitud, longitud)` o `(r, g, b)`).
   - Quieres proteger los datos contra modificaciones accidentales en funciones.
   - Quieres usar una secuencia como clave dentro de un diccionario.

3. **Usa `set` cuando:**
   - Necesitas eliminar duplicados de una colección de manera instantánea.
   - Vas a realizar frecuentes comprobaciones de pertenencia (`if x in coleccion:`).
   - Requieres calcular uniones, intersecciones o diferencias entre conjuntos de datos.

4. **Usa `dict` cuando:**
   - Quieres asociar un identificador significativo a datos (ej. `codigo_estudiantil -> datos_alumno`).
   - Modelas entidades complejas con múltiples atributos (similar a un objeto o JSON).
   - Creas tablas de lookup o índices en memoria para acelerar consultas.

## 5. 📝 Resumen de Métodos y Operaciones

### Métodos de Diccionarios (`dict`)
| Método | Propósito | Ejemplo |
| :--- | :--- | :--- |
| `d[key]` | Obtiene el valor asociado a la clave (lanza `KeyError` si no existe) | `nombre = estudiante["nombre"]` |
| `d.get(key, default)` | Obtiene el valor de forma segura sin romper el programa | `nota = estudiante.get("nota", 0.0)` |
| `d[key] = val` | Asigna o actualiza el valor para una clave | `estudiante["semestre"] = 6` |
| `d.keys()` | Devuelve una vista iterable con todas las claves | `for k in d.keys():` |
| `d.values()` | Devuelve una vista iterable con todos los valores | `for v in d.values():` |
| `d.items()` | Devuelve pares `(clave, valor)` para iterar | `for k, v in d.items():` |
| `d.update(otro_dict)` | Actualiza el diccionario insertando o sobreescribiendo pares | `d.update({"a": 1, "b": 2})` |
| `d.pop(key, default)` | Extrae y remueve una clave retornando su valor | `val = d.pop("clave", None)` |
| `d.clear()` | Elimina todos los elementos del diccionario | `d.clear()` |

### Métodos y Operadores de Conjuntos (`set`)
| Operación | Operador | Método | Descripción |
| :--- | :---: | :--- | :--- |
| **Pertenencia** | `in` | N/A | Verifica en tiempo instantáneo $O(1)$ si un elemento pertenece al set |
| **Agregar** | N/A | `s.add(elem)` | Añade un elemento al conjunto |
| **Eliminar seguro** | N/A | `s.discard(elem)` | Elimina el elemento sin fallar si no existe |
| **Eliminar estricto**| N/A | `s.remove(elem)` | Elimina el elemento (lanza `KeyError` si no existe) |
| **Unión** | `\|` | `s1.union(s2)` | Elementos en `s1`, en `s2` o en ambos |
| **Intersección** | `&` | `s1.intersection(s2)` | Elementos que están en ambos conjuntos |
| **Diferencia** | `-` | `s1.difference(s2)` | Elementos en `s1` pero no en `s2` |
| **Dif. Simétrica**| `^` | `s1.symmetric_difference(s2)` | Elementos en `s1` o en `s2`, pero no en ambos |
| **Subconjunto** | `<=` | `s1.issubset(s2)` | Verifica si todos los elementos de `s1` están en `s2` |
