## 1. Captura y conversión de datos (Casting)

Cuando usamos `input()` en Python, **todo lo que el usuario escribe se guarda como texto (`str`)**, sin importar si parece un número. Para poder operar matemáticamente con ese dato, debemos convertirlo explícitamente a otro tipo. A este proceso se le llama **casting**.

### Funciones de conversión más usadas

| Función | Convierte a | Ejemplo |
|---|---|---|
| `int()`   | Entero        | `int("25")` → `25` |
| `float()` | Decimal       | `float("3.14")` → `3.14` |
| `str()`   | Texto         | `str(10)` → `"10"` |
| `bool()`  | Booleano      | `bool(0)` → `False` |

⚠️ **Errores comunes:** si intentas convertir un texto que no representa un número (`int("hola")`), Python lanza un `ValueError`.

````python
# Ejemplo básico: sin conversión
edad = input("Ingresa tu edad: ")
print(type(edad))  # <class 'str'>, aunque el usuario escriba un número

# El siguiente cálculo fallaría si no convertimos:
# edad_futura = edad + 10  # ❌ TypeError: no se puede sumar str + int
````

````python
# Ejemplo correcto: con casting
edad_texto = input("Ingresa tu edad: ")
edad = int(edad_texto)  # conversión explícita a entero

edad_futura = edad + 10
print(f"Dentro de 10 años tendrás {edad_futura} años")
print(type(edad))  # <class 'int'>
````

````python
# Casting entre varios tipos
numero_str = "3.1416"

numero_float = float(numero_str)   # str -> float
numero_int   = int(numero_float)   # float -> int (trunca, no redondea)
numero_bool  = bool(numero_int)    # int  -> bool (0 = False, cualquier otro = True)

print(numero_str, type(numero_str))
print(numero_float, type(numero_float))
print(numero_int, type(numero_int))
print(numero_bool, type(numero_bool))
````

````python
# Manejo de errores de conversión con try/except
dato = "veinticinco"

try:
    valor = int(dato)
    print("Conversión exitosa:", valor)
except ValueError:
    print(f"No se puede convertir '{dato}' a entero")
````

## 2. Operadores relacionales

Los operadores relacionales (o de comparación) **comparan dos valores** y siempre devuelven un resultado booleano: `True` o `False`.

| Operador | Significado | Ejemplo | Resultado |
|---|---|---|---|
| `==` | Igual a | `5 == 5` | `True` |
| `!=` | Diferente de | `5 != 3` | `True` |
| `>`  | Mayor que | `7 > 10` | `False` |
| `<`  | Menor que | `7 < 10` | `True` |
| `>=` | Mayor o igual que | `5 >= 5` | `True` |
| `<=` | Menor o igual que | `4 <= 3` | `False` |


````python
a = 10
b = 20

print(f"a == b: {a == b}")
print(f"a != b: {a != b}")
print(f"a > b:  {a > b}")
print(f"a < b:  {a < b}")
print(f"a >= 10: {a >= 10}")
print(f"b <= 20: {b <= 20}")
````

## 3. Operadores lógicos

Los operadores lógicos permiten **combinar varias condiciones** (expresiones booleanas) en una sola.

| Operador | Significado | Se cumple cuando... |
|---|---|---|
| `and` | Y | **ambas** condiciones son verdaderas |
| `or`  | O | **al menos una** condición es verdadera |
| `not` | Negación | invierte el valor booleano |

### Tabla de verdad resumida

- `True and True` → `True`
- `True and False` → `False`
- `True or False` → `True`
- `False or False` → `False`
- `not True` → `False`

````python
edad = 20
tiene_matricula = True

# and: ambas condiciones deben cumplirse
puede_inscribirse = (edad >= 18) and tiene_matricula
print("¿Puede inscribirse?:", puede_inscribirse)

# or: basta con que una se cumpla
tiene_descuento = (edad < 18) or (edad > 60)
print("¿Tiene descuento?:", tiene_descuento)

# not: invierte el resultado
print("¿NO tiene matrícula?:", not tiene_matricula)
````

````python
# Combinando varios operadores relacionales y lógicos
nota = 3.8
asistencia = 85

aprueba = (nota >= 3.0) and (asistencia >= 80)
print(f"Nota: {nota}, Asistencia: {asistencia}% -> ¿Aprueba? {aprueba}")
````

## 4. Bloques de decisión: `if`, `elif`, `else`

Las estructuras condicionales permiten que el programa **tome decisiones** y ejecute distinto código según si una condición es verdadera o falsa.

### Sintaxis general

```python
if condicion_1:
    # se ejecuta si condicion_1 es True
elif condicion_2:
    # se ejecuta si condicion_1 es False y condicion_2 es True
else:
    # se ejecuta si ninguna condición anterior fue True
```

**Puntos clave:**
- La indentación (sangría) es obligatoria en Python; define qué código pertenece al bloque.
- Puede haber **cero o más** bloques `elif`.
- El bloque `else` es **opcional** y siempre es el último.
- Python evalúa las condiciones **en orden** y ejecuta solo el primer bloque cuyo resultado sea `True`.

````python
# Ejemplo simple: if / else
numero = 7

if numero % 2 == 0:
    print(f"{numero} es par")
else:
    print(f"{numero} es impar")
````

````python
# Ejemplo con if / elif / else: clasificar una nota
nota = 3.8

if nota >= 4.5:
    print("Desempeño superior")
elif nota >= 4.0:
    print("Desempeño alto")
elif nota >= 3.0:
    print("Desempeño básico (aprobado)")
else:
    print("Desempeño bajo (reprobado)")
````

````python
# Condicionales anidados y combinados con operadores lógicos
edad = int(input("Ingresa tu edad: "))
tiene_acompanante = True

if edad >= 18:
    print("Acceso permitido: eres mayor de edad")
elif edad >= 13 and tiene_acompanante:
    print("Acceso permitido con acompañante")
else:
    print("Acceso denegado")
````
