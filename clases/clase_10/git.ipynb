# Fundamentos de Git

---
## Apertura: el problema

### Una situación conocida

Carpetas con archivos como `trabajo_final.docx`, `trabajo_final_v2.docx`, `trabajo_final_v2_AHORA_SI.docx`. Código que dejó de funcionar sin saber qué se cambió, y sin poder volver atrás.

### El problema se agrava en equipo

- Varias personas editando los mismos archivos al mismo tiempo.
- Compartir código por USB, WhatsApp o correo genera copias desactualizadas.
- Si dos personas modifican el mismo archivo, alguien pierde su trabajo.

### La idea clave

Git resuelve estos problemas: funciona como una **máquina del tiempo** (recuperar cualquier versión anterior) y como una **herramienta de colaboración** (combinar el trabajo de varias personas de forma ordenada).

## Control de versiones y qué es Git

### ¿Qué es un sistema de control de versiones (VCS)?

Un sistema que **registra los cambios** hechos a un conjunto de archivos a lo largo del tiempo, de modo que se pueda recuperar cualquier versión anterior, ver quién cambió qué y por qué.

### ¿Qué es Git?

- Creado en 2005 por Linus Torvalds para desarrollar el kernel de Linux.
- Es **distribuido, rápido y pensado para proyectos grandes**.
- El historial es una **línea de tiempo de «fotografías»** del proyecto.

### Git ≠ GitHub

| Git | GitHub |
|---|---|
| Herramienta que corre en tu computador | Servicio web que aloja repositorios Git |
| Funciona sin internet | Necesita internet |
| Es el «motor» | Es la «plataforma social» alrededor del motor |

GitHub es a Git lo que Gmail es al correo electrónico: un servicio conveniente construido sobre una tecnología. Existen alternativas como GitLab y Bitbucket.

## Bloque 3 · Instalación y configuración

**Verificar la instalación:**

```bash
git --version
```

**Configuración inicial** (se hace una sola vez por máquina). Estos comandos modifican la configuración global del computador:

```bash
git config --global user.name "Tu Nombre"
git config --global user.email "tu_correo@ejemplo.com"
git config --global init.defaultBranch main
```

Git firma cada cambio con el nombre y el correo configurados, por eso es lo primero que se define. Para GitHub conviene usar el mismo correo de la cuenta.

**Comprobar la configuración:**

```bash
git config --list
```

## Modelo mental: las tres áreas

Este es **el concepto más importante** para entender Git.

```
Directorio de trabajo  --git add-->  Área de preparación  --git commit-->  Repositorio
(Working directory)                  (Staging area)                        (.git)
```

### Las tres áreas

1. **Directorio de trabajo:** los archivos que ves y editas en tu carpeta.
2. **Área de preparación (staging):** una «sala de espera» donde eliges *qué cambios* irán en la próxima fotografía.
3. **Repositorio:** el historial permanente de commits, guardado en la carpeta oculta `.git`.


### Estados de un archivo

- **Untracked:** Git no lo conoce todavía.
- **Modified:** fue cambiado pero no preparado.
- **Staged:** está listo para entrar al próximo commit.
- **Committed:** ya está guardado en el historial.

### ¿Por qué existe el staging?

Para poder hacer commits pequeños y con sentido, aunque se hayan modificado muchos archivos.

## Flujo básico: demo

Se recomienda ejecutar `git status` con frecuencia: muestra en qué estado está cada archivo y qué hacer a continuación.

### Preparación de la demo

Crea una carpeta limpia.

### Paso 1. Crear un repositorio

`git init` crea la carpeta oculta `.git`, donde vive todo el historial. Si se borra esa carpeta, el proyecto deja de ser un repositorio.

``` bash
mkdir mi-primer-repo
cd mi-primer-repo
git init -b main
git config user.name "Demo Clase"
git config user.email "demo@ejemplo.com"
```

*(`-b main` fija el nombre de la rama inicial; si ya se configuró `init.defaultBranch`, basta con `git init`.)*

### Paso 2. Revisar el estado

```bash
git status
```

### Paso 3. Crear un archivo y ver cómo lo detecta Git

El archivo nuevo aparece como **untracked**.

``` bash
echo "# Mi primer proyecto" > README.md
git status
```

### Paso 4. Preparar el cambio (staging)

Ahora el archivo aparece en «Changes to be committed».

``` bash
git add README.md
git status
```

### Paso 5. Hacer el primer commit

Un commit es una fotografía del proyecto con un mensaje que explica *qué* y *por qué* cambió. Los mensajes importan: con el tiempo, serán la forma de entender la historia del proyecto.

``` bash
git commit -m "Agrega README inicial"
```

### Paso 5.1. Ver el historial

```bash
git log
git log --oneline
```

### Paso 6. Modificar y revisar diferencias

`git diff` muestra lo que cambió y todavía no está en staging. Es buena costumbre revisarlo antes de hacer `git add`.

```bash
echo "Este proyecto es para aprender Git." >> README.md
git diff
```

```bash
git add README.md
git commit -m "Describe el propósito del proyecto"
git log --oneline
```

### Resumen de comandos del bloque

| Comando | Para qué sirve |
|---|---|
| `git init` | Crea un repositorio nuevo |
| `git status` | Muestra el estado actual |
| `git add <archivo>` | Prepara cambios para el commit |
| `git commit -m "msg"` | Guarda la fotografía en el historial |
| `git log --oneline` | Muestra el historial resumido |
| `git diff` | Muestra cambios sin preparar |

**Error común:** olvidar `git add` antes de `git commit`. El commit sale vacío o no incluye el archivo esperado. Se detecta con `git status`.

## Bloque 6 · Ramas y fusiones

### Concepto

Una rama es una línea de desarrollo independiente. Permite probar una idea nueva sin tocar el código que ya funciona: si la idea sirve, se une; si no, se descarta.

```
main:          A --- B --- C --------- F
                      \             /
nueva-funcion:         D --- E ----
```

- Una rama es simplemente **un puntero ligero a un commit**; crearlas es barato y rápido.
- `HEAD` indica en qué rama (y commit) se está parado ahora.

### Demo

```bash
# Lista las ramas (la actual lleva un asterisco)
git branch
```

```bash
# Crea y cambia a una rama nueva
git switch -c nueva-funcion
echo "Nueva sección" >> README.md
git add README.md
git commit -m "Agrega nueva sección al README"
cat README.md
```

Al volver a `main`, el contenido del archivo cambia solo: cada rama tiene su propia versión.

```bash
# Fusiona la rama en main y borra la rama ya fusionada
git merge nueva-funcion
git branch -d nueva-funcion
git log --oneline --graph
```

```bash
git switch main
cat README.md
```

## Repositorios remotos y GitHub

### Concepto

Un **remoto** es una copia del repositorio alojada en otro lugar (por ejemplo, GitHub). Por convención, el remoto principal se llama `origin`.

```
Tu computador (local)  <--- push / pull --->  GitHub (remoto)
```

### Comandos clave

| Comando | Qué hace |
|---|---|
| `git clone <url>` | Descarga una copia completa de un repositorio |
| `git remote add origin <url>` | Conecta el repositorio local con uno remoto |
| `git push` | Envía los commits al remoto |
| `git pull` | Trae y fusiona los commits nuevos del remoto |
| `git fetch` | Trae los cambios del remoto sin fusionarlos |

### Ejemplo con GitHub (en la terminal, con cuenta propia)

Estos pasos requieren una cuenta y credenciales, por eso no se ejecutan desde el cuaderno.

1. Crear un repositorio vacío en GitHub (sin README).
2. Conectarlo y subir el proyecto:

```bash
git remote add origin https://github.com/usuario/mi-primer-repo.git
git push -u origin main
```

3. Editar el README desde la web de GitHub, hacer commit allí y luego traer el cambio:

```bash
git pull
```

**Nota sobre autenticación:** GitHub ya no acepta contraseña normal por HTTPS; se usa un token de acceso personal, el *Git Credential Manager* o llaves SSH. Conviene revisar la documentación vigente de GitHub.

### Ciclo de trabajo diario

`clone` se hace una sola vez; después el ciclo es: `pull` → trabajar → `add` → `commit` → `push`.
