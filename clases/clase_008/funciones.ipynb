## 1. ¿Qué es una función y por qué usarla?

Una **función** es un bloque de código con nombre que realiza una tarea específica y que podemos **ejecutar cuantas veces queramos**.

La **programación modular** consiste en dividir un programa en piezas pequeñas (funciones), donde cada una hace **una sola cosa**.

**Ventajas**
- **Reutilización:** se escribe una vez y se usa muchas veces.
- **Legibilidad:** el programa se lee como una serie de pasos con nombre.
- **Mantenimiento:** si hay que corregir algo, se cambia en un solo lugar.
- **Pruebas:** cada pieza se puede probar por separado.

`print()`, `len()`, `input()`, `int()`, `max()`...

## 2. Sintaxis básica: definir y llamar una función

```python
def nombre_funcion(parametros):
    # cuerpo de la función (indentado)
    instrucciones
```

- `def` es la palabra reservada que **define** la función.
- El **nombre** sigue las reglas de las variables (usa `snake_case`).
- Los **paréntesis** contienen los parámetros (pueden estar vacíos).
- Los **dos puntos** `:` son obligatorios.
- El **cuerpo** debe ir **indentado**.

Definir una función **no la ejecuta**; solo la crea. Para ejecutarla hay que **llamarla**.

### 2.1 Una función sin parámetros

````python
def saludar():
    print("¡Hola, bienvenido al curso!")

# # Llamadas a la función
saludar()
saludar()
````

### 2.2 Una función con un parámetro

````python
def saludar_persona(nombre):
    print(f"¡Hola, {nombre}!")

saludar_persona(7)
saludar_persona("Luis")
````

## 3. Parámetros y argumentos

- **Parámetro:** la variable que aparece en la **definición** de la función.
- **Argumento:** el valor concreto que se envía en la **llamada**.

````python
def presentar(nombre, edad):      # nombre y edad son PARÁMETROS
    print(f"{nombre} tiene {edad} años")

presentar("Marta", 20)            # "Marta" y 20 son ARGUMENTOS
````

### 3.1 Argumentos posicionales

````python
# El orden importa
presentar("Marta", 20)
presentar(20, "Marta")   # sintácticamente válido, pero el resultado no tiene sentido
````

### 3.2 Argumentos nombrados (*keyword arguments*)

````python
# Con nombre, el orden ya no importa
presentar(edad=20, nombre="Marta")
````

### 3.3 Valores por defecto

````python
def saludar(nombre, saludo="Hola"):
    print(f"{saludo}, {nombre}")

saludar("Ana")                      # usa el valor por defecto
saludar("Ana", "Buenos días")       # sobrescribe el valor por defecto
saludar("Ana", saludo="Qué tal")
````

> **Regla:** los parámetros con valor por defecto deben ir **después** de los que no lo tienen.

### 3.4 Error común: número incorrecto de argumentos

````python
presentar("Marta")      # falta un argumento
````

## 4. Funciones con retorno (`return`)

Hasta ahora las funciones solo **mostraban** algo con `print`. Una función con retorno **calcula un valor y lo devuelve** a quien la llamó, para poder guardarlo o reutilizarlo.

```
def nombre(parametros):
    ...
    return resultado
```

### 4.1 Primer ejemplo

````python
def area_rectangulo(base, altura):
    return base * altura

a = area_rectangulo(5, 3)       # el valor devuelto se guarda en 'a'
# print("Área:", a)

# El valor devuelto se puede usar en expresiones
print("Doble del área:", 2 * area_rectangulo(5, 3))
````

### 4.2 `print` vs `return`

| `print()` | `return` |
|---|---|
| Muestra un valor en pantalla | Devuelve un valor al código que llamó |
| El valor **no** se puede reutilizar | El valor **sí** se puede reutilizar |
| La función continúa | La función **termina** en ese punto |

````python
def suma_con_print(a, b):
    print(a + b)

def suma_con_return(a, b):
    return a + b

r1 = suma_con_print(2, 3)     # muestra 5, pero no devuelve nada
r2 = suma_con_return(2, 3)    # no muestra nada, pero devuelve 5

print("r1 =", r1)
print("r2 =", r2)
````

### 4.3 Sin `return`, la función devuelve `None`
`None` es el valor especial de Python que significa "nada".

````python
def saludar(nombre):
    print(f"Hola, {nombre}")

resultado = saludar("Ana")
print(resultado)
````

### 4.4 Retornar varios valores
Se separan con comas. Python devuelve una **tupla** que podemos separar en varias variables.

````python
def estadisticas(numeros):
    minimo = min(numeros)
    maximo = max(numeros)
    promedio = sum(numeros) / len(numeros)
    return minimo, maximo, promedio

datos = [4, 8, 15, 16, 23, 42]
mn, mx, prom = estadisticas(datos)


print(f"mín={mn}, máx={mx}, promedio={prom:.2f}")
````

### 4.5 `return` termina la función
Cualquier código después de un `return` ejecutado **no se ejecuta**. Esto sirve para salir antes en casos especiales.

````python
def dividir(a, b):
    if b == 0:
        return "No se puede dividir entre cero"



    print("algo")
    return a / b

# print(dividir(10, 0))
print(dividir(10, 2))
````

### 4.6 Ejemplo: funciones que devuelven `True` o `False`

````python
def es_par(n):
    return n % 2 == 0

def es_mayor_de_edad(edad):
    return edad >= 18

print(es_par(4), es_par(7))
print(es_mayor_de_edad(17), es_mayor_de_edad(20))

# Se usan directamente en un if
numero = 10
if es_par(numero):
    print(f"{numero} es par")
````

## 5. Funciones lambda

Una **función lambda** es una función **pequeña y sin nombre**, escrita en una sola línea. Se usa para operaciones simples.

```
lambda parametros: expresion
```

- No lleva `def` ni `return`: **devuelve automáticamente** el resultado de la expresión.
- Solo puede contener **una expresión** (no varias instrucciones).

### 5.1 Comparación `def` vs `lambda`

````python
# Con def
def cuadrado(x):
    return x ** 2

# Con lambda (equivalente)
cuadrado_lambda = lambda x: x ** 2

print(cuadrado(4))
print(cuadrado_lambda(4))
````

### 5.2 Lambdas con varios parámetros o sin parámetros

````python
suma = lambda a, b: a + b
print(suma(3, 5))

area = lambda base, altura: base * altura / 2
print(area(6, 4))

saludo = lambda: "¡Hola!"
print(saludo())

# También admiten valores por defecto
potencia = lambda base, exp=2: base ** exp
print(potencia(5), potencia(2, 10))
````

### 5.4 ¿Para qué sirven en la práctica?
Lo más habitual es usarlas **como argumento de otra función** cuando necesitamos una regla rápida y de un solo uso.

**a) Ordenar con `sorted()` y el parámetro `key`**

````python
palabras = ["python", "sol", "programación", "ia", "datos"]


# palabras_ordenadas = sorted(palabras, key=lambda p: 5)
# Ordenar por longitud de la palabra
# print(palabras_ordenadas)

# Ordenar una lista de tuplas por el segundo elemento
estudiantes = [("Ana", 4.5), ("Luis", 3.2), ("Luis", 4.2), ("Marta", 4.9)]

por_nombre = sorted(estudiantes, key=lambda e: e[0])
organizado = sorted(por_nombre, key=lambda e: e[1])

print(organizado)
````

**b) Transformar con `map()`**: aplica una función a cada elemento

````python
numeros = [1, 2, 3, 4, 5]
cuadrados = list(map(lambda x: x + 3, numeros))
print(cuadrados)
print(numeros)
````

**c) Filtrar con `filter()`**: se queda con los elementos que cumplen la condición

````python
numeros = [1, 2, 3, 4, 5, 6, 7, 8]
pares: list[int] = list(filter(lambda x: x % 2 == 1, numeros))
print(pares)
````

### 5.5 ¿Cuándo usar `def` y cuándo `lambda`?

| Usa `lambda` cuando... | Usa `def` cuando... |
|---|---|
| La operación es **muy simple** (una expresión) | La lógica tiene **varias líneas** o condiciones complejas |
| Se usa **una sola vez**, normalmente como argumento | Se usa **en varios lugares** |
| No necesita docstring | Quieres **nombre claro y documentación** |

> Si la lambda se vuelve difícil de leer, conviértela en una función con `def`.

## 6. Alcance de variables (*scopes*)

El **alcance** (*scope*) indica **en qué parte del programa una variable existe y puede usarse**.

- **Variable local:** se crea **dentro** de una función; solo existe allí.
- **Variable global:** se crea **fuera** de las funciones; se puede leer desde cualquier parte.

### 6.1 Variables locales

````python
def calcular():
    valor_local = 10 * 2     # variable LOCAL
    print("Dentro:", valor_local)

calcular()

print(valor_local)        # no existe fuera de la función
````

### 6.2 Variables globales

````python
iva = 0.19                    # variable GLOBAL

def precio_con_iva(precio):
    return precio * (1 + iva)  # leer una global está permitido

print(precio_con_iva(100000))
````

### 6.3 Variable local con el mismo nombre que una global
Si dentro de la función **asignamos** un valor a un nombre, Python crea una variable **local nueva**; la global no cambia.

````python
x = "global"

def f():
    x = "local"
    print("dentro de f:", x)

f()
print("fuera de f:", x)
````

### 6.4 Los parámetros también son variables locales

````python
def duplicar(n):
    n = n * 2
    return n

numero = 5
print(duplicar(numero))
print(numero)                 # no cambió
````

### 6.5 Modificar una global: la palabra clave `global`
Por defecto, no se puede **modificar** una global desde una función:

````python
contador = 0

def incrementar():
    contador = contador + 1   # Python la interpreta como local y falla

incrementar()
````

Con `global` se le indica a Python que use la variable del exterior:

````python
contador = 0

def incrementar():
    global contador
    contador += 1

incrementar()
contador = ""
incrementar()
print(contador)
````

### 6.6 Buena práctica: evitar `global`
Usar `global` hace que la función **dependa de algo externo** y sea difícil de seguir y probar. Lo recomendable es **recibir datos por parámetros y entregar resultados con `return`**.

````python
def incrementar(valor):
    return valor + 1

contador = 0
contador = incrementar(contador)
contador = incrementar(contador)
print(contador)
````

## 7. Desacoplamiento de componentes

Cuando escribimos funciones queremos que sean **independientes** entre sí:

- **Acoplamiento:** qué tanto **depende** una función de otras o de variables externas. Queremos que sea **bajo**.
- **Cohesión:** qué tan enfocada está una función en **una sola tarea**. Queremos que sea **alta**.

> Una función desacoplada solo necesita sus **parámetros** y entrega su resultado con **`return`**.

### 7.1 Ejemplo acoplado

````python
# Una sola función pide datos, calcula y muestra.
def procesar_pedido():
    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))
    total = precio * cantidad * 1.19
    print(f"Total a pagar: ${total:,.0f}")

procesar_pedido()    # requiere interacción
````

Problemas:
- No se puede **reutilizar** el cálculo en otro programa.
- No se puede **probar** automáticamente.
- Si cambia el IVA o el formato, hay que modificar toda la función.

### 7.2 Ejemplo desacoplado

````python
def calcular_subtotal(precio, cantidad):
    return precio * cantidad

def calcular_total(subtotal, tasa_iva=0.19):
    return subtotal * (1 + tasa_iva)

def formatear_moneda(valor) -> str:
    return f"${valor:,.0f}"


precio: int = 25000
cantidad = 3
subtotal = calcular_subtotal(precio, cantidad)
total = calcular_total(subtotal)
print("Total a pagar:", formatear_moneda(total))
````

### 7.3 Separar responsabilidades
Una forma sencilla de organizar un programa es separarlo en tres tipos de funciones:

| Tipo | Responsabilidad | Ejemplo |
|---|---|---|
| **Entrada** | Obtener datos | `input()`, leer un archivo |
| **Lógica** | Procesar / calcular | `calcular_total()`, `promedio()` |
| **Salida** | Mostrar resultados | `print()`, `formatear_moneda()` |

### 7.4 Señales de alto acoplamiento
- Uso de muchas variables **globales**.
- Funciones que mezclan **cálculo + `input` + `print`**.
- Cambiar una función obliga a modificar varias más.
- No se puede probar una función sin ejecutar todo el programa.

## 9. Resumen

| Tema | Idea clave |
|---|---|
| **Sintaxis** | `def nombre(parametros):` + cuerpo indentado. Definir no ejecuta; hay que **llamar** la función. |
| **Parámetros** | Posicionales, nombrados y con valor por defecto. |
| **Retorno** | `return` devuelve un valor reutilizable y termina la función. Sin `return` se obtiene `None`. Se pueden devolver varios valores. |
| **Lambda** | `lambda params: expresion`. Función corta y anónima, útil con `sorted`, `map` y `filter`. |
| **Scopes** | Local (dentro de la función) vs global (fuera). Evita `global`; usa parámetros y `return`. |
| **Desacoplamiento** | Funciones pequeñas, con una sola tarea, que no dependen de globales ni de `input`/`print`. |

> **Regla práctica:** una función recibe lo que necesita por **parámetros**, entrega su resultado con **`return`** y hace **una sola cosa**.
