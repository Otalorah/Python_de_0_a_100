# Secuencias indexadas: listas y tuplas

## 1. Crear una lista

Una secuencia es una colección ordenada. Las listas usan corchetes `[]` y el primer índice es `0`.

```python
colores = ['rojo', 'verde', 'azul']
print('Secuencia:', colores)
print('Cantidad:', len(colores))
```

## 2. Índices positivos y negativos

Usamos `secuencia[indice]`. El índice `-1` representa el último elemento y `-2` el penúltimo.

```python
frutas = ['manzana', 'pera', 'banana']
print('Primera:', frutas[0])
print('Segunda:', frutas[1])
print('Última:', frutas[-1])
```

## 3. Slicing

La forma `secuencia[inicio:fin:paso]` permite extraer una parte. El inicio se incluye y el fin no.

```python
letras = ['a', 'b', 'c', 'd', 'e']
print('Parte central:', letras[1:4])
print('Desde el inicio:', letras[:3])
print('De dos en dos:', letras[0:5:2])
```

## 4. Mutabilidad de las listas

Las listas son mutables: podemos cambiar un elemento después de crearlas.

```python
temperaturas = [18, 20, 22]
print('Antes:', temperaturas)
temperaturas[1] = 21
print('Después:', temperaturas)
```

## 5. Tuplas e inmutabilidad

Las tuplas usan paréntesis `()`. Se pueden leer y cortar, pero no cambiar. La línea comentada produciría `TypeError`.

```python
coordenadas = (10, 20)
print('Primer valor:', coordenadas[0])
print('Slicing:', coordenadas[:1])
# coordenadas[0] = 99
```

## 6. Métodos de listas

`append` agrega al final, `pop` quita y devuelve, `remove` quita por valor y `sort` ordena.

```python
tareas = ['estudiar', 'practicar']
tareas.append('repasar')
quitada = tareas.pop()
print('Se quitó:', quitada)
print('Tareas:', tareas)
```

```python
colores = ['rojo', 'azul', 'verde', 'azul']
colores.remove('azul')
print('Después de remove:', colores)

puntajes = [80, 65, 95, 70]
puntajes.sort()
print('Después de sort:', puntajes)
```

## 7. Operaciones entre listas y tuplas

Podemos combinar secuencias con `+`, repetirlas con `*`, comprobar si contienen un elemento con `in` y convertir una lista en tupla o una tupla en lista.

```python
lista_a = [1, 2]
lista_b = [3, 4]
colores = ('rojo', 'azul')

print('Listas combinadas:', lista_a + lista_b)
print('Lista repetida:', lista_a * 2)
print('¿Está azul en la tupla?:', 'azul' in colores)

lista_desde_tupla = list(colores)
tupla_desde_lista = tuple(lista_a)
print('Tupla convertida en lista:', lista_desde_tupla)
print('Lista convertida en tupla:', tupla_desde_lista)
```

## 8. Recorrer una lista con `for`

Un bucle `for` repite un bloque para cada elemento de una secuencia. Así podemos procesar todos los valores sin escribir una línea para cada uno.

```python
frutas = ['manzana', 'pera', 'banana']

for fruta in frutas:
    print('Fruta:', fruta)
```

## 9. Operaciones entre conjuntos (`set`)

Los conjuntos se escriben con llaves `{}` y no permiten elementos repetidos. No tienen índices, pero permiten comparar grupos de elementos:

- `|` o `.union()`: unión de ambos conjuntos.
- `&` o `.intersection()`: elementos que aparecen en los dos.
- `-` o `.difference()`: elementos del primer conjunto que no están en el segundo.
- `in`: comprueba si un elemento pertenece al conjunto.

```python
grupo_a = {'Ana', 'Luis', 'Marta'}
grupo_b = {'Luis', 'Marta', 'Pablo'}

print('Unión:', grupo_a | grupo_b)
print('Intersección:', grupo_a & grupo_b)
print('Diferencia de A:', grupo_a - grupo_b)
print('¿Ana pertenece al grupo B?:', 'Ana' in grupo_b)
```
